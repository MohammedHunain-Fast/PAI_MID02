```python
import pandas as pd
```

```python
df = pd.read_csv("data.csv")
df.head()
```

```python
df.dtypes
```

```python
df.columns = df.columns.str.lower().str.replace(" ","_")
df.head()
```

```python
df = df.map(lambda x: x.lower().replace(' ', '_') if isinstance(x, str) else x)

df.head()
```

```python
df.dtypes
```

```python
df['msrp'] = df['msrp'].astype('Int64')
df['engine_hp'] = df['engine_hp'].astype("Int64")
```

```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(19, 6))
plt.plot( df['msrp'], df['engine_hp'], color = 'red', marker = 'o', label = 'name')
plt.title('MSRP vs Engine HP')
plt.xlabel('MSRP')
plt.ylabel('HP')
plt.grid(True, linestyle = "--", alpha = 0.5)
plt.tight_layout()
plt.show()
```

```python
df.dtypes
```

```python
plt.figure(figsize=(19, 6))
plt.bar( df['year'], df['msrp'], color = 'red', edgecolor = 'blue', width= 1)
plt.title('MSRP vs Year')
plt.xlabel('Year')
plt.ylabel('MSRP')
plt.grid(True, linestyle = "--", alpha = 0.5)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(19, 6))
plt.barh( df['year'], df['msrp'], color = 'red')
plt.title('MSRP vs Year')
plt.xlabel('Year')
plt.ylabel('MSRP')
plt.legend()
plt.grid(True, linestyle = "--", alpha = 0.5)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(19, 6))
plt.scatter( df['year'], df['msrp'], color = 'blue', alpha = 1)
plt.title('MSRP vs Year')
plt.xlabel('Year')
plt.ylabel('MSRP')
plt.legend()
plt.grid(True, linestyle = "--", alpha = 0.5)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize = (19, 6))
plt.title('Popularity')
plt.hist(df['popularity'], color = 'black')
plt.legend(df['popularity'])
plt.grid(True, linestyle = '--', alpha = 0.5)
plt.tight_layout()
plt.show()
```

```python
import seaborn as sns
```

```python
sns.set_theme(style = "whitegrid")
sns.lineplot(data = df, x = 'msrp', y = 'engine_hp', marker = 'o', color = 'blue')
plt.xlabel('MSRP')
plt.title('MSRP vs HP')
plt.ylabel('Engine HP')
plt.tight_layout()
plt.show()
```

```python
sns.set_theme(style = 'white')
sns.scatterplot(data = df, x = 'popularity', y = 'msrp', color = 'red')
plt.title('P VS M')
plt.xlabel('Popularity')
plt.ylabel('MSRP')
plt.tight_layout()
plt.show()
```

```python
sns.set_theme(style = 'whitegrid')
sns.regplot(data = df, x = 'popularity', y = 'msrp', color = 'red')
plt.title('P VS M')
plt.xlabel('Popularity')
plt.ylabel('MSRP')
plt.tight_layout()
plt.show()
```

```python
correlation = df.corr(numeric_only=True)
sns.set_theme(style = 'white')
sns.heatmap(correlation, annot=True, cmap = 'coolwarm', vmin = 1, vmax = 1, fmt = '.2f')
plt.title('P VS M')
plt.xlabel('Popularity')
plt.ylabel('MSRP')
plt.tight_layout()
plt.show()
```
